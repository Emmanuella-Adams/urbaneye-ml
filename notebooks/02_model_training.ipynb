# UrbanEye — Notebook 02: PyTorch U-Net Model Training

**Objective:** Instantiate the PyTorch U-Net segmentation architecture, train over multiple epochs with BCE + Dice Loss, record training/validation loss history, save loss curve figures, and serialize trained model weights to `model_weights.pth`.

```python
import sys
import os
import torch

# Add src to path
sys.path.append(os.path.abspath('../src'))
from data_utils import create_dataloaders
from model_utils import create_model, train_unet_model, save_model
from eval_utils import plot_training_curves
```

```python
# 1. Initialize PyTorch DataLoaders
img_dir = '../sample_data/images'
mask_dir = '../sample_data/masks'
train_loader, val_loader, test_loader = create_dataloaders(img_dir, mask_dir, batch_size=4)
print("DataLoaders successfully initialized.")
```

```python
# 2. Instantiate PyTorch U-Net Architecture
device = 'cuda' if torch.cuda.is_available() else 'cpu'
model = create_model('unet', in_channels=3, out_channels=1)
print(f"U-Net model initialized on device: {device}")
```

```python
# 3. Train U-Net Model with BCE + Dice Loss
model, history = train_unet_model(model, train_loader, val_loader, epochs=10, lr=1e-3, device=device)
```

```python
# 4. Plot & Save Training Loss Curves
plot_training_curves(history, save_path='../figures/training_loss_curve.png')
```

```python
# 5. Save Model Checkpoint Weights
save_model(model, filepath='../model_weights.pth')
```
