# UrbanEye — Notebook 01: Data Preparation & Tile Verification

**Objective:** Verify 256x256 satellite RGB imagery tiles and matching building footprint binary masks, verify data normalization, and instantiate train (70%), validation (15%), and test (15%) PyTorch DataLoaders.

```python
import sys
import os
import torch
import matplotlib.pyplot as plt

# Add src to path
sys.path.append(os.path.abspath('../src'))
from data_utils import BuildingFootprintDataset, create_dataloaders
```

```python
# 1. Load full dataset & check counts
img_dir = '../sample_data/images'
mask_dir = '../sample_data/masks'
dataset = BuildingFootprintDataset(img_dir, mask_dir)
print(f"Total satellite tile pairs found: {len(dataset)}")
```

```python
# 2. Visualize sample image and binary mask pair
image_tensor, mask_tensor = dataset[0]
print("Image Tensor Shape:", image_tensor.shape)  # Should be (3, 256, 256)
print("Mask Tensor Shape:", mask_tensor.shape)    # Should be (1, 256, 256)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(image_tensor.numpy().transpose(1, 2, 0))
axes[0].set_title("Satellite RGB Tile (256x256)")
axes[0].axis('off')

axes[1].imshow(mask_tensor.squeeze().numpy(), cmap='gray')
axes[1].set_title("Building Footprints Ground Truth Mask")
axes[1].axis('off')
plt.show()
```

```python
# 3. Create PyTorch DataLoaders (Train 70%, Val 15%, Test 15%)
train_loader, val_loader, test_loader = create_dataloaders(
    image_dir=img_dir, mask_dir=mask_dir, batch_size=4
)
print(f"Train Batches: {len(train_loader)} | Val Batches: {len(val_loader)} | Test Batches: {len(test_loader)}")
```
